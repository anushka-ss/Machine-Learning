```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
df = pd.DataFrame({
    'cat_col1':['A','B','A','C','B','A','C','B','A','C'],
    'cat_col2':['X','Y','X','Z','Y','X','Z','Y','X','Z'],
    'label':    ['class0','class1','class2','class0','class1','class2','class0','class1','class2','class0']
})
```

```python
x= df.drop(columns='label')
y=df['label']
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(
    x,
    y,
    train_size=0.8,
    random_state=42
)
```

```python
encoder = LabelEncoder()
encoder.fit_transform(xtrain['cat_col1'])
```
