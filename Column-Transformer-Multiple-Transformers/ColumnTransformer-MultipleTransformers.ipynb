```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler,OrdinalEncoder
from category_encoders import TargetEncoder
from sklearn.compose import ColumnTransformer
```

```python
df = pd.DataFrame({
    'Age': [22, 35, 28, 41, 30],                  # Numerical (small range)
    'Salary': [25000, 85000, 42000, 120000, 67000],  # Numerical (large range)
    'Customer_ID': ['ID001', 'ID025', 'ID050', 'ID075', 'ID100'],  # Part of 100 unique values
    'Purchased': ['Yes', 'No', 'Yes', 'No', 'Yes'],  # 2 unique values
    'Experience': [1.5, 8.0, 3.2, 15.0, 5.5]         # Numerical
})

print(df)
```

```python
# x = df.copy()
# y = df['Experience']   # or any other target column
```

```python
x = df.drop(columns='Experience')  # Features
y = df['Experience']
```

```python
print(x.columns)
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(
    x,
    y,
    train_size=0.8,
    random_state=42
)
```

```python
preprocessing = ColumnTransformer(
    transformers=[
        ('scaling',MinMaxScaler(),['Age','Salary']),
        ('target',TargetEncoder(),['Customer_ID']),
        ('ordinal',OrdinalEncoder(),['Purchased'])],

    remainder='passthrough' ) # Keep the remaining columns unchanged
```

```python
xtrain
```

```python
preprocessing.fit_transform(xtrain,ytrain) 
```
