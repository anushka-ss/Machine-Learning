```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,RobustScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,f1_score,precision_score,accuracy_score
from sklearn.impute import SimpleImputer
from sklearn.metrics import classification_report
```

```python
df=pd.read_csv('loan_default_risk_dataset.csv')
```

```python
x=df.drop(columns='Loan_Default_Risk')
y=df['Loan_Default_Risk']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
sns.boxplot(x)
```

```python
sns.boxplot(x['Monthly_Savings'])
```

```python
num_pipeline=Pipeline(
    steps=[
        ('imputer',SimpleImputer(strategy='mean')),
        ('rscaler',RobustScaler())
    ]
)
```

```python
num_pipeline1=Pipeline(
    steps=[
        ('imputer',SimpleImputer(strategy='mean')),
        ('sscaler',StandardScaler())
    ]
)
```

```python
preprocessing=ColumnTransformer(
    transformers=[
        ('rcsaler',num_pipeline,['Debt_Amount']),
        ('sscaler',num_pipeline1,['Retirement_Age','Monthly_Savings'])
    ]
)
preprocessing
```

```python
main_pipeline=Pipeline(
    steps=[
        ('pre1',preprocessing),
        ('model',LogisticRegression(penalty=None,solver='lbfgs'))
    ]
)

main_pipeline.fit(xtrain,ytrain)
```

# model evaluation

```python
ytrain_pred=main_pipeline.predict(xtrain)
ytest_pred=main_pipeline.predict(xtest)
```

```python
ytrain.value_counts()
```

# confusion matrix for train :

```python
confusion_matrix(ytrain,ytrain_pred)
```

# confusion matrix for test :


```python
confusion_matrix(ytest,ytest_pred)
```
