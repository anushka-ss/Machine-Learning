```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import RobustScaler
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.metrics import f1_score,confusion_matrix,precision_score,recall_score,accuracy_score
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report
```

# Model Training

```python
df=pd.read_csv('loan_data.csv')
```

```python
x=df.drop(columns='loan_status')
y=df['loan_status']
```

```python
x.info()
```

```python
num_cols=x.select_dtypes(include='number').columns
cat_cols=x.select_dtypes(include='str').columns
df[cat_cols].nunique()
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
num_cols
cat_cols
```

```python
sns.boxplot(df[num_cols])
```

# Conclusion

numerical colomns are having  different range values  and more outliers  so we need to apply Robust Scalewr to scale the values

```python
df[cat_cols]
```

```python
preprocessing=ColumnTransformer(
    transformers=[
        ('scaler',RobustScaler(),num_cols),
        ('encoder1',OneHotEncoder(handle_unknown='ignore'),['person_gender', 'person_home_ownership','loan_intent', 'previous_loan_defaults_on_file']),
        ('encoder2',OrdinalEncoder(),['person_education'])
    ]
)
preprocessing
```

```python
df.columns
```

```python
main_pipeline=Pipeline(
    steps=[
        ('pre',preprocessing),
        ('model',LogisticRegression(penalty=None,solver='lbfgs'))# penalty = l2(ridge) by default
    ]
)
main_pipeline.fit(xtrain,ytrain)
```

# Model Evaluation

```python
ytrain.value_counts()
```

# Confusion matrics

```python
ytrain_pred=main_pipeline.predict(xtrain)
ytest_pred=main_pipeline.predict(xtest)
```
